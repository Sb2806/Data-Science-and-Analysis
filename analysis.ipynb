```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
pio.templates.default = "plotly_white"
```

```python
data = pd.read_csv("D:\VS code\Sample_ Superstore.csv\Sample_ Superstore.csv", encoding = 'latin')

data
```

```python
data.isna().sum()
```

```python
data.duplicated().values.any()
```

```python
data.describe()
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'], format = 'mixed')
data['Ship Date'] = pd.to_datetime(data['Ship Date'], format = 'mixed')
```

```python
data['Order Year'] = data['Order Date'].dt.year
data['Order Month'] = data['Order Date'].dt.month
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.info()
```

```python
data.shape
```

```python
monthly_sales_trends = data.groupby('Order Month')['Sales'].sum().reset_index()

monthly_sales_trends
```

```python
plt.figure(figsize = (10, 5))
sns.lineplot(
    data = monthly_sales_trends,
    x = 'Order Month',
    y = 'Sales',
    marker = 'o',
    color = 'steelblue')
plt.title("Monthly Sales Analysis", fontsize = 14, pad = 13)
plt.xlabel("Order Month", fontsize = 12)
plt.ylabel("Sales", fontsize = 12)
plt.xticks(rotation = 45, ha = 'right')

plt.tight_layout()
plt.show()
```

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
categories = sales_by_category['Category']
sales = sales_by_category['Sales']

fig, ax = plt.subplots(figsize = (5, 5))

wedges, texts, autotexts = ax.pie(
    sales,
    labels = categories,
    autopct = '%1.1f%%',
    startangle = 90,
    colors = plt.cm.Pastel1.colors,
    pctdistance = 0.7
)

plt.setp(autotexts, size = 10, weight = "bold", color = "black")
plt.setp(texts, size = 10)

ax.set_title("Sales Analysis by Category", fontsize = 18, pad = 20)
plt.tight_layout()
plt.show()
```

```python
sales_by_subcategory = data.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_subcategory
```

```python
plt.figure(figsize = (10, 5))
sns.barplot(
    data = sales_by_subcategory,
    x = 'Sub-Category',
    y = 'Sales')

plt.title("Sales Analysis by Sub Category", fontsize = 16, pad = 15)
plt.xlabel("Sub Category", fontsize=12)
plt.ylabel("Sales", fontsize = 12)
plt.xticks(rotation =45, ha ='right')

plt.tight_layout()
plt.show()
```

```python
profit_by_month = data.groupby('Order Month')['Profit'].sum().reset_index()
profit_by_month
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(
    data = profit_by_month,
    x = 'Order Month',
    y = 'Profit',
    marker = 'o',
    color = 'steelblue')
plt.title("Monthly Profit Analysis", fontsize = 14, pad=13)
plt.xlabel("order Month", fontsize=12)
plt.ylabel("Profit", fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()
profit_by_category
```

```python
categories = profit_by_category['Category']
profit = profit_by_category['Profit']

fig, ax = plt.subplots(figsize = (5, 5))

wedges, text, autotexts = ax.pie(
    profit,
    labels = categories,
    autopct = '%1.1f%%',
    colors = plt.cm.Pastel1.colors,
    pctdistance = 0.7)
plt.setp(autotexts, size = 10, weight = "bold", color = "black")
plt.setp(text, size = 10)

ax.set_title("Profit Analysis by Category", fontsize = 18, pad = 20)
plt.tight_layout()
plt.show()
```

```python
profit_by_subcategory = data.groupby('Sub-Category')['Profit'].sum().reset_index()
profit_by_subcategory
```

```python
plt.figure(figsize = (10, 5))
sns.barplot(
    data = profit_by_subcategory,
    x = 'Sub-Category',
    y = 'Profit',
)
plt.title("Profit Analysis by Sub Category", fontsize = 16, pad = 15)
plt.xlabel("Sub Category", fontsize = 12)
plt.xticks(rotation = 45, ha = 'right')

plt.tight_layout()
plt.show()
```

```python
profit_by_region = data.groupby('Region')['Profit'].sum().reset_index()
profit_by_region
```

```python
plt.figure(figsize = (10, 5))
sns.barplot(
    data = profit_by_region,
    x = 'Region',
    y = 'Profit',
)

plt.title("Profit Analysis by Region", fontsize = 16, pad = 15)
plt.xlabel("Region", fontsize = 12)
plt.ylabel("Profit", fontsize = 12)
plt.xticks(rotation = 45, ha = 'right')
plt.tight_layout()
plt.show()
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
df_melted = sales_profit_by_segment.melt(
    id_vars = 'Segment',
    value_vars = ['Sales', 'Profit'],
    var_name = 'Metric',
    value_name = 'Amount'
)
plt.figure(figsize = (8, 6))
sns.barplot(
    data = df_melted,
    x = 'Segment',
    y = 'Amount',
    hue = 'Metric',
    palette = 'pastel'
)
plt.title("Sales and Profit Analysis by Customer Segment", fontsize = 16, pad = 15)
plt.xlabel("Customer Segments", fontsize = 12)
plt.ylabel("Amount", fontsize = 12)
plt.legend(title = "Metric")

plt.tight_layout()
plt.show()
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```

```python
segments = sales_profit_by_segment['Segment']
ratio = sales_profit_by_segment["Sales_to_Profit_Ratio"]

fig, ax = plt.subplots(figsize = (5, 5))

wedges, texts, autotexts = ax.pie(
    ratio,
    labels = segments,
    autopct = '%1.1f%%',
    startangle = 90,
    colors = plt.cm.Pastel1.colors,
    pctdistance = 0.7
)
plt.setp(autotexts, size = 10, weight = "bold", color = "black")
plt.setp(texts, size = 10)

ax.set_title("Sales and Profit Ratio by Segment", fontsize = 18, pad = 20)
plt.tight_layout()
plt.show()
```
