**We can see here we have some NaN values in our dataset. We need to solve this by using data preprocessing method.**

```python
numerical_cols = data.select_dtypes(include = np.number).columns
data[numerical_cols] = data[numerical_cols].fillna(data[numerical_cols].mean())

categorical_cols = data.select_dtypes(include = 'object').columns
for col in categorical_cols:
  mode_val = data[col].mode()[0] if not data[col].mode().empty else np.nan
  data[col] = data[col].fillna(mode_val)

print("Missing values after filling:")
display(data.isna().sum())


```

**We have cleared the NaN values with some techniques. For numerical columns, missing (NaN) values are placed with the mean of each respective. This ensures that the missing values don't distort the dataset, while keeping the distribution relatively stable. For each categorical column the mode is computed. If mode exist, missing values are filled with this most frequent category. If no mode exists, the code assign NaN. After filling missing values in both numerical and categorical columns, the code checks how many are left. This helps confirm whether the imputation was successful.**

**1. What is the monthly sales of the store and identify which month had the highest sales and which month had the lowest sales?**

```python
monthly_sales_trends =  data.groupby('Order Month')['Sales'].sum().reset_index()

monthly_sales_trends
```

```python
fig = px.line(monthly_sales_trends,
              x = 'Order Month',
              y = 'Sales',
              title = 'Monthly Sales Analysis')
fig.show()
```

**Ans-1) Sales begin moderately, drop sharply in 2nd month, and then recover significantly in the 3rd. Month 4 through 8 are flat with minor fluctuations. Sales picked up strongly in the final quarter, hitting their highest levels of the year, especially in month 11 and 9. The pick come in month 11, while the lowest point was back in month 2. The graph clearly shows a seasonal pattern in sales, with a slump early in the year followed by steady growth leading up to the end of the year. Likely boosted by holiday or seasonal demand.**

**7) What are the sales and profit by customer segment?**

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
color_palatte = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(x = sales_profit_by_segment['Segment'],
                     y = sales_profit_by_segment['Sales'],
                     name = 'Sales',
                     marker_color = color_palatte[0]))

fig.add_trace(go.Bar(x = sales_profit_by_segment['Segment'],
                     y = sales_profit_by_segment['Profit'],
                     name = 'Profit',
                     marker_color = color_palatte[1]))

fig.update_layout(title = 'Sales and Profit Analysis by Customer Segment',
                  xaxis_title = 'Customer Segment', yaxis_title = 'Amount')
fig.show()
```

**Ans-7) Customer Segment has the hoghest sales. Also generates the highest profit, through profit is much smaller compared to sales. Corporate Segment has moderate sales. The profit is samller than consumers but still noticeble. Home office Segment has the lowest sales. Also shows the lowest profit among the three segments.**

**8) What are the sales to profit ratio?**



```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales']  / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```

```python
fig = px.pie(sales_profit_by_segment,
             values = 'Sales_to_Profit_Ratio',
             names = 'Segment',
             hole = 0.2,
             color_discrete_sequence = px.colors.qualitative.Pastel)
fig.update_traces(textposition = 'inside', textinfo = 'percent+label')
fig.update_layout(title_text = 'Sales and Profit Ratio by Segment', title_font = dict(size = 18))
fig.show()
```

**Ans-8) We can see here, the consumer segment holds the leargest share of the sales and profit ratio at 36.7%. This indicates consumers contribute the most overall. The corporate segment accounts for 32.8%, making it the second-leargest contributor. The home office segment represents 30.5%,slightly lower than corporate. While smaller, it still holds a significant portion.**
